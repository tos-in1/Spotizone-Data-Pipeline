### Table  Creation: dim_date

#### Create Table

```python
%sql
USE CATALOG spotizone;
USE SCHEMA gold;

DROP TABLE IF EXISTS spotizone.gold.dim_date;

CREATE TABLE spotizone.gold.dim_date(
  month_start DATE,
  date_key STRING,
  year INT,
  month_name STRING,
  month_short_name STRING,
  month_number INT,
  quarter INT,
  year_quarter STRING
);
```

#### Insert into dim_date

```python
%sql
--  Truncate table first before inserting
TRUNCATE TABLE spotizone.gold.dim_date;

INSERT INTO spotizone.gold.dim_date
WITH full_date AS (
  SELECT explode(
    sequence(
      to_date('2025-01-01'),
      to_date('2025-12-31'),
      interval 1 month
    )
  ) AS date_range
)

SELECT
  date_range AS month_start,
  date_format(date_range, 'yyyyMMdd') AS date_key,
  year(date_range) AS year,
  date_format(date_range, 'MMMM') AS month_name,
  date_format(date_range, 'MMM') AS month_short_name,
  month(date_range) AS month_number,
  quarter(date_range) AS quarter,
  concat(year(date_range), '-Q', quarter(date_range)) AS year_quarter
FROM full_date;
```
